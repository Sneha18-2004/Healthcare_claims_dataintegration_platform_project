```python
dbutils.widgets.text("partition_num", "0", "Partition Number (0-3)")

partition_num = dbutils.widgets.get("partition_num")

print(f"Partition to truncate: {partition_num}")
```

```python
truncate_table = f"{BRONZE_PARTITION_TABLE}_{partition_num}"

print(f"Truncating: {truncate_table}")

spark.sql(f"TRUNCATE TABLE {truncate_table}")

print(f"✓ Truncated: {truncate_table}")
```

```python
verify_count = spark.table(truncate_table).count()

print(f"{'='*60}")
print(f"TRUNCATE COMPLETE")
print(f"{'='*60}")
print(f"Table  : {truncate_table}")
print(f"Records: {verify_count:,} (should be 0)")
print(f"{'='*60}")

if verify_count == 0:
    print("✓ Verification PASSED: Table is empty")
    print("✓ Ready for next file")
else:
    print(f"⚠ WARNING: Table still has {verify_count:,} records")
```
